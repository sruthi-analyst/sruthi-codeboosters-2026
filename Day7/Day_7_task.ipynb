```python
!pip install chromadb sentence-transformers -q
print("Installed dependencies!")
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
import chromadb
print("Imported libraries!")
print(f"ChromaDB version: {chromadb.__version__}")
```

```python
new_df = pd.read_csv("notes_dataset.csv")
print(df.duplicated().sum())
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

print("Model loaded successfully!")
print(f"Model produces vectors of size: {model.get_sentence_embedding_dimension()}")
```

```python
doc_embeddings = model.encode(new_df['content'].tolist())
print("Embedding shape:", doc_embeddings.shape)
print()
print("Sample embedding(embedding of row 0):")
print(doc_embeddings[0][:10])
print()
print(f"Min value of row 0: {doc_embeddings[0].min()}")
print(f"Max value of row 0: {doc_embeddings[0].max()}")
print()
print("Document Embeddings generated successfully!")
```

```python
query = "Quantum Mechanics"
query_embedding = model.encode(query)
print("Query embedding generated successfully!")
```

```python
# Calculate cosine similarities
# sentence_transformers.util.cos_sim returns a tensor of similarities
print("~="*30)
print(f"\tSEMANTIC SEARCH for '{query}'")
print("~="*30)

from sentence_transformers import util
similarities_tensor = util.cos_sim(query_embedding, doc_embeddings)[0]
print(f"Data type of cos_sim returns: {similarities_tensor.dtype}")

print(f"Query: '{query}'\n")
print("Semantic Search Results (ranked by similarity):")
threshold = 0.16
found = []
for i in range(len(similarities_tensor)):
  if(similarities_tensor[i] > threshold):
    flag = "FOUND"
    found.append(i)
    print(f"  Score: {similarities_tensor[i]:.4f} - {flag} - [doc_{i}]: {new_df['topic'][i]} - {new_df['content'][i]}")

print("\nRelevant Documents Found:")
for i in found:
  print(f'Doc_{i}', end=" ")
```

```python
chroma_client =chromadb.Client()
collection=chroma_client.get_or_create_collection("notes_dataset")
print("ChromaDB client created (in-memory mode)")

print(f"Collection name:notes_dataset ")
print(f"Documents in collection :{collection.count()}")
```

```python
sample_docs = new_df['content'].tolist()
sample_ids = new_df['note_id'].tolist()
sample_metadata = []

# Correctly generate metadata for each document
for index, row in new_df.iterrows():
    sample_metadata.append({
        'subject': row['subject'],
        'topic': row['topic']
    })
sample_embeddings = model.encode(sample_docs)

print(f"Sample docs (first 2): \n\t{sample_docs[:2]}")
print(f"Sample ids (first 2): \n\t{sample_ids[:2]}")
print(f"Sample metadata (first 2): \n\t{sample_metadata[:2]}")
print(f"Sample embeddings (first 2): \n\t{sample_embeddings[0][1]}")

collection.add(
    documents=sample_docs,
    embeddings=sample_embeddings, # Pass the generated embeddings
    ids=sample_ids,
    metadatas=sample_metadata
)

print("\n\nDocuments added successfully")
print(f"Number of documents in collection: {collection.count()}")
```

```python
query="Explain thermodynamics and mechanics?"

results=collection.query(
    query_texts=[query],
    n_results=3
)

print("RESULT KEYS AVAILABLE:")
print(list(results.keys()))
```

```python
print(f"Query :'{query}'")
print("="*60)
print()

matched_docs = results['documents'][0]
matched_ids = results['ids'][0]
matched_distance = results['distances'][0]
matched_metadata = results['metadatas'][0]

for rank,(doc , doc_id,dist,meta)in enumerate(zip(matched_docs,matched_ids,matched_distance,matched_metadata)):
  print(f"Rank:{rank}|ID:{doc_id}|Distance:{dist:.2f}")
  print(f"Subject:{meta['subject']}|Topic:{meta['topic']}")
  print(f"Document:{doc}\n")
```

```python
filtered_results = collection.query(
    query_texts=[query],
    n_results=3,
    where={"topic": "Thermodynamics"}
)
print(f"FILTERED QUERY: {query}")
print("Filter only Thermodynamics documents")
print("-~"*25)

for rank, (doc, dist, meta) in enumerate(zip(
    filtered_results['documents'][0],
    filtered_results['distances'][0],
    filtered_results['metadatas'][0]),start=1):
  print(f"Rank: {rank} | Distance: {dist:.2f} | Subject: {meta['subject']}")
  print(f"{doc}\n")
```
